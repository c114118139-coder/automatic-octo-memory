## Mission 4：變數、運算與學習分數公式

建立學習相關變數：

- `study_hours`
- `sleep_hours`
- `screen_time_hours`
- `attendance_rate`

並依照題目提供的概念建立一個學習分數：

**學習分數 = 50 + 3 × 睡眠時間 + 2 × 學習時間 − 1.2 × 螢幕時間 + 0.15 × 出席率**

最後限制分數在 0～100 之間。


```python
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

raw_score = (
    50
    + 3 * sleep_hours
    + 2 * study_hours
    - 1.2 * screen_time_hours
    + 0.15 * attendance_rate
)

score = max(0, min(100, raw_score))

print("學習分數：", score)
```

```python
# Verification
assert 0 <= score <= 100
print("Verification：分數已限制在 0～100。")
```

## Mission 5：函式與條件判斷

使用 `def`、參數、`return`、`if / elif / else` 建立分數描述函式。


```python
def describe_score(score):
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到及格標準"
    else:
        return "需要加強"

print(describe_score(85))
print(describe_score(70))
print(describe_score(50))
```

```python
# Verification
assert describe_score(85) == "表現良好"
assert describe_score(70) == "達到及格標準"
assert describe_score(50) == "需要加強"
print("Verification：函式與 if / elif / else 判斷正常。")
```

## Mission 7：建立 60 筆學生 DataFrame

建立 60 位學生的學習資料，欄位包含：

- `student_id`
- `department`
- `sleep_hours`
- `screen_time_hours`
- `study_hours`
- `attendance_rate`

資料範圍依題目要求：
- 睡眠時間平均約 6.7 小時、標準差約 1.1
- 螢幕時間平均約 5.0 小時、標準差約 1.5，限制在 1～10
- 學習時間平均約 2.5 小時、標準差約 1.0，限制在 0.5～6
- 出席率平均約 88、標準差約 8，限制在 60～100


```python
n = student_count

student_ids = np.arange(1, n + 1)

department_choices = np.array(["資管系", "資工系", "工管系"])
department = rng.choice(department_choices, size=n)

sleep_hours = rng.normal(loc=6.7, scale=1.1, size=n)
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n)
study_hours = rng.normal(loc=2.5, scale=1.0, size=n)
attendance_rate = rng.normal(loc=88, scale=8, size=n)

sleep_hours = np.round(sleep_hours, 1)
screen_time_hours = np.round(np.clip(screen_time_hours, 1, 10), 1)
study_hours = np.round(np.clip(study_hours, 0.5, 6), 1)
attendance_rate = np.round(np.clip(attendance_rate, 60, 100), 1)

df = pd.DataFrame({
    "student_id": student_ids,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

df.head()
```

```python
# Verification
assert len(df) == 60
assert list(df.columns) == [
    "student_id",
    "department",
    "sleep_hours",
    "screen_time_hours",
    "study_hours",
    "attendance_rate"
]
assert df["student_id"].min() == 1
assert df["student_id"].max() == 60
assert df["screen_time_hours"].between(1, 10).all()
assert df["study_hours"].between(0.5, 6).all()
assert df["attendance_rate"].between(60, 100).all()

print("Verification：DataFrame 已建立，共", len(df), "筆資料。")
```

## Mission 8：計算 quiz_score

依照前面的公式計算每位學生的 `quiz_score`。

另外加入 0～5 的固定亂數，讓每位學生有一些自然差異：

**quiz_score = 50 + 3 × sleep_hours + 2 × study_hours − 1.2 × screen_time_hours + 0.15 × attendance_rate + random(0, 5)**

最後使用 `clip()` 將分數限制在 0～100。


```python
bonus = rng.uniform(0, 5, size=n)

df["quiz_score"] = (
    50
    + 3 * df["sleep_hours"]
    + 2 * df["study_hours"]
    - 1.2 * df["screen_time_hours"]
    + 0.15 * df["attendance_rate"]
    + bonus
).clip(0, 100).round(1)

df[
    [
        "student_id",
        "study_hours",
        "sleep_hours",
        "screen_time_hours",
        "attendance_rate",
        "quiz_score"
    ]
].head()
```

```python
# Verification
assert "quiz_score" in df.columns
assert df["quiz_score"].between(0, 100).all()
print("Verification：quiz_score 已建立，且所有分數都在 0～100。")
```

## Mission 9：用 `.loc` 製造缺失值與異常值

題目指定：

- `student_id` 6（索引 5）的 `sleep_hours` 設為缺失值
- `student_id` 18（索引 17）的 `sleep_hours` 設為缺失值
- `student_id` 9（索引 8）的 `screen_time_hours` 設為 15，製造異常值

使用 `.loc` 與 `np.nan` 完成。


```python
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0

print("student_id 6、18 的 sleep_hours：")
print(df.loc[[5, 17], ["student_id", "sleep_hours"]])

print("\nstudent_id 9 的 screen_time_hours：")
print(df.loc[8, ["student_id", "screen_time_hours"]])
```

```python
# Verification
assert pd.isna(df.loc[5, "sleep_hours"])
assert pd.isna(df.loc[17, "sleep_hours"])
assert df.loc[8, "screen_time_hours"] == 15.0
print("Verification：缺失值與異常值已成功建立。")
```

## Mission 11：條件篩選與資料分析

完成以下分析：

1. 找出高分學生（`quiz_score >= 80`）
2. 找出睡眠時間缺失的學生
3. 找出螢幕時間異常（`> 10`）的學生
4. 找出學習時間至少 3 小時的學生
5. 找出 `quiz_score >= 80` 的學生

練習使用布林條件與 DataFrame 篩選。


```python
# 1. quiz_score >= 80
high_score_condition = df["quiz_score"] >= 80
high_score_students = df[high_score_condition]

print("=== quiz_score >= 80 ===")
display(high_score_students)

# 2. sleep_hours 缺失
missing_sleep_students = df[df["sleep_hours"].isna()]
print("=== sleep_hours 缺失 ===")
display(missing_sleep_students)

# 3. screen_time_hours > 10
abnormal_screen_students = df[df["screen_time_hours"] > 10]
print("=== screen_time_hours > 10 ===")
display(abnormal_screen_students)

# 4. study_hours >= 3
active_study_students = df[df["study_hours"] >= 3]
print("=== study_hours >= 3 ===")
display(active_study_students)

# 5. quiz_score >= 80 的 student_id
print("=== 高分學生 student_id ===")
print(high_score_students["student_id"].tolist())
```

```python
# Verification
assert len(high_score_students) == (df["quiz_score"] >= 80).sum()
assert len(missing_sleep_students) == 2
assert len(abnormal_screen_students) >= 1
assert (active_study_students["study_hours"] >= 3).all()
print("Verification：五種條件篩選皆完成。")
```
